```python
import json
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# JSON 파일 읽기
with open('training_weights.json', 'r') as f:
    data = json.load(f)

# "W_out"을 numpy array(2D)로 변환
weights = np.array(data["W_out"])

# 컬러맵(heatmap)으로 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(weights, cmap="coolwarm", annot=False)  # 또는 cmap="coolwarm", "plasma" 등
plt.title("W_out Weight Matrix Color Map")
plt.xlabel("Output Neuron Index")
plt.ylabel("Residual Neuron Index")
plt.show()
```

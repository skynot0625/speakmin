```python
import json
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# init_taus.json 및 training_weights.json 파일 읽기
with open('training_weights.json', 'r') as f:
    data = json.load(f)
with open('init_taus.json', 'r') as f:
    taus_dict = json.load(f)
    taus = np.array(taus_dict["tau"])  # key에 맞게 인덱싱

# W_out weight matrix
weights = np.array(data["W_out"])

# tau 값 오름차순 정렬 인덱스 획득
sorted_indices = np.argsort(taus)

# tau 및 가중치 행렬 정렬
sorted_weights = weights[sorted_indices, :]
sorted_taus = taus[sorted_indices]

# 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(sorted_weights, cmap="coolwarm", annot=False)
plt.title("W_out Weight Matrix Sorted by Tau Values")
plt.xlabel("Output Neuron Index")
plt.ylabel("Residual Neuron Index (sorted by tau)")
plt.show()

```

```python
import json
import numpy as np
import matplotlib.pyplot as plt

# JSON 파일 읽기
with open('training_weights.json', 'r') as f:
    data = json.load(f)

# "W_out"을 numpy array(2D)로 변환
weights = np.array(data["W_out"])

# 1D로 평탄화
weights_flat = weights.flatten()

# 히스토그램으로 분포 시각화
plt.figure(figsize=(8, 6))
plt.hist(weights_flat, bins=50, color='blue', alpha=0.7)
plt.title('Distribution of W_out Weights')
plt.xlabel('Weight Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

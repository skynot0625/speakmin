```python
import numpy as np
from scipy.stats import truncnorm
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)

num_neu_res = 400  # 예시

loc = 0.1                # mean
scale = 0.1         # std = 0.05
a, b = -1, 1             # clip 범위

a_trunc = (a - loc) / scale
b_trunc = (b - loc) / scale

W_res = truncnorm.rvs(
    a_trunc, b_trunc,
    loc=loc, scale=scale,
    size=(num_neu_res, num_neu_res),
    random_state=rng
)

print(W_res.mean(), W_res.std())

# 히스토그램 + weight map 시각화
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# 1) 히스토그램
axes[0].hist(W_res.flatten(), bins=50, density=True, alpha=0.7)
axes[0].set_title("Truncated normal weights")
axes[0].set_xlabel("weight")
axes[0].set_ylabel("density")

# 2) weight map (앞 30x30만 보기)
im = axes[1].imshow(W_res[:30, :30], cmap="coolwarm", vmin=-1, vmax=1)
axes[1].set_title("Weight map (30x30)")
plt.colorbar(im, ax=axes[1])

plt.tight_layout()
plt.show()
```
